```python
from sklearn.linear_model import Lasso
import numpy as np
x=np.array([[1,25,160],[2,30,165],[3,35,170]])
y=np.array([[30000],[40000],[50000]])
model=Lasso(alpha=0.1)
model.fit(x,y)
print(model.coef_)
```
