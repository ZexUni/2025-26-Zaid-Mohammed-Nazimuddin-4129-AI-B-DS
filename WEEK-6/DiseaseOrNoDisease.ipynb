```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
data={
    "age":[25,45,356,50,23],
    "bp":[120,140,130,150,110],
    "disease":[0,1,1,1,0]
}
df=pd.DataFrame(data)
X=df[["age","bp"]]
Y=df["disease"]
model=LogisticRegression()
model.fit(X,Y)
print("Disease Prediction:", model.predict([[40,135]]))
```
