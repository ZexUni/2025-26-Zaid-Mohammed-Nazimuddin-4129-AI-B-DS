```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
data={
    "X":[1,2,3,4,5],
    "Y":[2,4,5,4,5]
}
df=pd.DataFrame(data)
X=df[["X"]]
Y=df["Y"]
model=LinearRegression()
model.fit(X,Y)
print("Intercept:",model.intercept_)
print("Slope:",model.coef_)
df["Y_pred"]=model.predict(X)
print(df)
```
